## 3. Basic operations

## 4. Indexing, slicing and iterating

การเข้าถึงข้อมูล: 

arr[i],  ตำแหน่งของข้อมูล ใน index ที่ i 

arr[i,j]  สร้าง Array จากจาก array เดิม ในตำแหน่งข้อมูลที่ i ถึง ก่อน j 

การ slicing: arr[1:4], arr[:,1]

การวนลูปด้วย for

```python
a = np.arange(10)
b = np.array([[1,2,3],[4,5,6]])
print("a: ", a)
print("b: ",b)   # print("b: \n",b)  
```

```python
# Slice array index:2 to index: 6  [2,3,4,5,6]
print(a[2:7])  # slice
# slice row 
for row in b:
    print(row)
```

## 5. Shape manipulation 

Changing the shape of an array

```python
arr = np.arange(12)
print(arr)
```

```python
# 3 row 4 columns 
print(arr.reshape(3,4))  
# 4 row 3 columns 
print(arr.reshape(4,3) )
# 6 rows 2 columns 
print(arr.reshape(6,2) )

# Transpose Array 
print(arr.reshape(6,2).T)
```

## 6. Stacking together different arrays

np.vstack: ต่อ Array ในแนวตั้ง

np.hstack: ต่อ Array ในแนวนอน 


```python
a = np.ones((2,2))
b = np.zeros((2,2))
print(a)
print(b)
```

```python
print(np.vstack((a,b))) # ต่อแนวตั้ง  V: vertical
print(np.hstack((a,b))) # ต่อแนวนอน h: horizontal 
```
