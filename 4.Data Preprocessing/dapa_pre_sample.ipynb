```python
# Manage Outlier 
# หาสาเหตุ 
# delete outliner 
Q1 = titanic['Fare'].quantile(0.25)
Q3 = titanic['Fare'].quantile(0.75)
IQR = Q3 - Q1
print( IQR)

# กำหนดขอบเขต (IQR method)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

```

```python
# lower 
titanic['Fare'] >= lower
# Upper 
titanic['Fare'] <= upper 
# print(titanic['Fare'] <= upper)
# print(titanic['Fare'] >= lower)
titanic_no_outlier = titanic[  (titanic['Fare'] >= lower) & (titanic['Fare'] <= upper )  ]
# titanic_no_outlier = titanic[ (titanic['Fare'] >= lower) & (titanic['Fare'] <= upper) ]
titanic_no_outlier
```
