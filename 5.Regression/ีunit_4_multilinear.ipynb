```python
# setup 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
df = pd.read_csv('datasets/advertising.csv', index_col=0)
df.head()
```

```python
sns.pairplot(df, x_vars=['TV','Radio','Newspaper'], y_vars='Sales', height=3.5,kind='reg')
plt.show()
```

```python
g = sns.pairplot(df)
g.fig.set_size_inches(9, 9)
```

```python

# Multi linear regression 
feature_cols = ['TV','Radio','Newspaper']
X = df[feature_cols]   # X พิมพ์ใหญ่ 
y = df.Sales
```

```python
X.shape
type(y)
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print(model.coef_)
print(model.intercept_)  # m and b
list(zip(feature_cols, model.coef_.round(2) ))
```

```python
# correlation coefcient 
df.corr().round(3)
```

```python
# R-Squared
model.score(X, y)
```

```python
# Predict 
x_input = [[300, 0, 0], [0, 300, 0], [0, 0, 300], [100, 200, 200], [100, 200, 0]]
model.predict(x_input)
```

```python
# Error 
y_pred = model.predict(X)
mae =mean_absolute_error(y,y_pred)
print(round(mae, 3))  # กรณีเป็น Floating ต้องใช้รูปแบบนี้ครับ 
mse = mean_squared_error(y,y_pred) 
print(f"MSE: {mse:.4f}")
rmse=np.sqrt(mse)
print(f"RMSE: {rmse:.4f}")
```
